#Leitura de um arquivo.xlsx

```python
#leitura de um arquivo.xlsx

enem = pd.read_excel('/content/Enem_2024_Amostra_Perfeita.xlsx')
```

```python
display(enem)
```

```python
# Filtrar o DataFrame 'enem' para incluir apenas alunos do Distrito Federal
df_distrito_federal = enem[enem['SG_UF_PROVA'] == 'DF']

# Exibir as primeiras linhas do novo DataFrame para verificar o filtro
display(df_distrito_federal)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Gráfico de barras com a variável TP_COR_RACA
plt.figure(figsize=(10, 6))
sns.countplot(data=enem, x='TP_COR_RACA', palette='viridis')
plt.title('Distribuição por Cor/Raça dos Participantes do ENEM')
plt.xlabel('Cor/Raça')
plt.ylabel('Número de Participantes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# 2. Selecionar alunos com NOTA_MT_MATEMATICA > 680
alunos_matematica_alta = enem[enem['NOTA_MT_MATEMATICA'] > 680]
frequencia_uf = alunos_matematica_alta['SG_UF_PROVA'].value_counts().reset_index()
frequencia_uf.columns = ['SG_UF_PROVA', 'Frequencia']

print("Tabela de Distribuição de Frequência por UF para Alunos com NOTA_MT_MATEMATICA > 680:")
display(frequencia_uf)
```
