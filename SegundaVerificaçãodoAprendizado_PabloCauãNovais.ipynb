```python
#Questão 2.Desenvolva um programa que dada uma nota ele indicará a menção correta.
nota = float(input())

if nota >= 9:
    print("SS")
elif nota >= 7:
    print("MS")
elif nota >= 5:
    print("MM")
elif nota >= 3:
    print("MI")
elif nota >= 0:
    print("II")
```

```python
#Questão 3.Dados 3 números quaisquer identifique qual o maior e o menor número.
numero1 = int(input())
numero2 = int(input())
numero3 = int(input())

maior_numero = max(numero1, numero2, numero3)
menor_numero = min(numero1, numero2, numero3)
print("O maior número é:", maior_numero)
print("O menor número é:", menor_numero)
```

```python
#Questão 4.Considere a seguinte lista de notas em um trecho de código em Python. Calcule quantos alunos obtiveram nota suficiente para aprovação (nota maior ou igual a 5.0)
notas = [4.5, 7.0, 3.0, 8.5, 5.0, 9.0, 7.0, 9.5, 10.0, 9.5]
alunos_aprovados = sum(nota >= 5.0 for nota in notas)
print("Número de alunos aprovados:", alunos_aprovados)
```

```python
#Questão 5.Implemente um programa em Python que leia um valor inicial N e imprima a sequência de valores do cálculo de N! (N fatorial) e o seu resultado. Use N=5 para testar o seu algoritmo. Exemplo: 5! = 5 x 4 x 3 x 2 x 1 = 120
n = int(input())
fatorial = 1
for i in range(1, n + 1):
    fatorial *= i
print(f"{n}! = {fatorial}")
```

```python
#Questão 6.Faça um programa Python que imprime a quantidade de números pares de 100 até 200, incluindo-os.
contador = 0

for i in range(100, 201):
    if i % 2 == 0:
        contador += 1

print(contador)
```

```python
#Questão 7.Faça um programa Python que calcule a soma dos 100 primeiros números inteiros maiores que 1 cujas raízes quadradas são exatas.
soma = 0

for i in range(2, 102):
    soma += i ** 2

print(soma)
```

```python
#Questão 8.Faça um programa Python que leia o arquivo “enem_2024_amostra_perfeita” e faça um gráfico de barras com a variável tp_sexo e um gráfico de pizza com a variável cor_raca.
#usei a ia pra ajudar a fazer o gráfico de pizza.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
enem = pd.read_excel('/content/Enem_2024_Amostra_Perfeita (2).xlsx')

#Gráfico de barras
plt.figure(figsize=(10, 6))
sns.countplot(data=enem, x='TP_Sexo', hue = 'TP_Sexo', palette='deep', legend = False)
plt.title('Distribuição por sexo dos Participantes do ENEM')
plt.xlabel('Sexo')
plt.ylabel('Número de Participantes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

#Gráfico de pizza
dados_raca = enem['TP_COR_RACA'].value_counts()
plt.figure(figsize=(8, 8))
cores = sns.color_palette('pastel')[0:len(dados_raca)]

plt.pie(dados_raca,
        labels=dados_raca.index,
        autopct='%1.1f%%',
        startangle=140,
        colors=cores)
plt.title('Distribuição por Cor/Raça no ENEM')
plt.show()




```
